```python
nhf.to_csv('cleaned_nigeriahealthfacilities.csv', index=False)
population_data.to_csv('cleaned_population_data.csv', index=False)
```

```python
# Loading the cleaned data into PostgreSQL database for further analysis using SQL queries 

from sqlalchemy import create_engine


engine = create_engine("postgresql+psycopg2://postgres:245074@localhost:5432/facility-to-population_ratio_analysis")

try:
    engine
    print('connection successful')
except:
    print('unable to connect')
```

```python
#exporting data to postgreSQL

nhf.to_sql(name= 'nigeriahealthfacilities', con=engine, if_exists='replace', index=False)
population_data.to_sql(name= 'populationdata', con=engine, if_exists='replace', index=False )
```
